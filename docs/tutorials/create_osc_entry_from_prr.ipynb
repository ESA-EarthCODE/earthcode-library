# Create an OSC entry from the PRR.

In order to create an OSC entry from an existing PRR collection you'll have to either:
1. Create a new project and then a new product
2. Create a new product and add it to an existing project

This tutorial shows how to use the two earthcode-library functions to handle each case.

## 1. Create a new project and then a new product

```python
# import all the relevant functions from the earthcode-library.
import json
from pathlib import Path
from typing import Any, Mapping, Sequence
from urllib.request import urlopen

from earthcode.git_add import (
    save_product_collection_to_catalog,
    save_project_collection_to_osc,
)

from earthcode.prr import osc_product_from_prr, osc_project_from_prr

from earthcode.validator import validate_catalog
```

```python
catalog_root = Path('../../../open-science-catalog-metadata/')
```

```python
# define the prr collection url for the creation of the project and create a json object
prr_collection_url = "https://eoresults.esa.int/stac/collections/mwd-lstm-sim" #pasquale
with urlopen(prr_collection_url) as response:
  prr_collection = json.load(response)

prr_collection
```

### Define all the project variables missing from the PRR and create the project collection to the OSC

```python
project_id = "sup-fcp-multisensor-water-discharges"
project_status = "ongoing"
technical_officer = (
      "Elisabetta Lamboglia",
      "elisabetta.lamboglia@esa.int",
  )
consortium_members = [
      ("Planetek Italia", "info@planetek.it"),
      ("National Technical University of Athens", "")
  ]
project_start_datetime = '2025-01-01'
project_end_datetime = '2027-12-31'

project_bbox = [[6.6, 36.6, 18.6, 47.1]]
```

```python
project = osc_project_from_prr(
    prr_collection,  
    project_id,
    project_status,
    technical_officer,
    consortium_members,
    project_start_datetime,
    project_end_datetime,
    project_bbox)
```

```python
save_project_collection_to_osc(project, catalog_root)
```

```python
validate_catalog(catalog_root)
```

### Define all the product variables missing from the PRR and create the product collection to the OSC

```python
prr_collection_url = "https://eoresults.esa.int/stac/collections/mwd-chime-sim"
with urlopen(prr_collection_url) as response:
  prr_collection = json.load(response)

prr_collection
```

```python
product_missions=[
      "prisma",
      "enmap",
]

product_variables=[
      "discharge",
      "water-leaving-spectral-radiance",
      "surface-water-chemistry",
]
```

```python
# create osc product collection from PRR collection
product = osc_product_from_prr(prr_collection, prr_collection_url, product_missions, product_variables, project_id)
```

```python
# save the product to the catalog
save_product_collection_to_catalog(product, catalog_root)
```

```python
validate_catalog(catalog_root)
```
